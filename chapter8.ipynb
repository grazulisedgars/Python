```python
# Another python native library time. Also does not need to be installed
import time
t1 = time.time()
print(t1) # Provides time in seconds since 1970s

print(datetime.datetime.fromtimestamp(t1)) # This line translates time to human friendly output

t2 = time.time() * 1000 # We would provide time in miliseconds
print(t2)
print(datetime.datetime.fromtimestamp(t2 / 1000)) # For this to work we need to convert time stamp back to seconds
```

```python
d2 = '2026-05-18 09:35:02'
d2_datetime = datetime.datetime.strptime(d2,'%Y-%m-%d %H:%M:%S')
print(d2_datetime)
print(d2_datetime.timestamp()) # With this line we get back to second format. In case we need to transform it the other way aroun
```

```python
import pandas as pd
df = pd.read_csv('BTC-USDT.csv')
df
```

```python
print(df['date'].dtype)
df['date'] = pd.to_datetime(df['date'], unit ="ms")
df
```

```python
df['weekday'] = df['date'].dt.weekday # We created a new column weekday. Which shows an integer of a weekday
df
```

```python
# Mapping

day_name_map = {
    0: 'Monday',
    1: 'Tuesday',
    2: 'Wednesday',
    3: 'Thursday',
    4: 'Friday',
    5: 'Saturday',
    6: 'Sunday'
}

df['weekday'] = df['weekday'].map(day_name_map)
df
```

```python
df['date_diff'] = df['date'] - df['date'].shift(1) # shift(1) shift the value by 1
df
```

```python
df['date_diff'].value_counts()
```

```python
df.groupby('weekday').agg({'volume':'mean'}) # We see that on weekend volumes are down in trading cryptos.
```

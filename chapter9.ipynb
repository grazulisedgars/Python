```python
import pandas as pd
```

```python
df= pd.read_csv('BTC-USDT.csv')
df
```

```python
df['date'] = pd.to_datetime(df['date'], unit ="ms")
df
```

```python
df.set_index(df['date'],inplace=True)
del df['date']
df
```

```python
# Creating new columns for Moving Averages
# 50 moving average based on close price. We want to do computation of rolling period 50. Computing mean the average price

df['ma20'] = df['close'].rolling(20).mean()
df['ma50'] = df['close'].rolling(50).mean()
df
```

```python
df['buy_signal'] = False
df['sell_signal'] = False
df.loc[df['ma20']> df['ma50'], 'buy_signal'] =True # Moving Average cross over
df.loc[df['ma20']< df['ma50'], 'sell_signal'] = True # Moving Average cross over
df
```

```python
df['sell_signal'].value_counts()
```

```python
# Backtesting

balance = 1000
position = None

for index, row in df.iterrows():    # Loop going to return 2 values 
    # buy condition
    if position is None and row['buy_signal'] is True:
        buy_price = row['close']
        usd_size = balance          # How much we are going to buy
        position = {
            'buy_price': buy_price,
            'usd_size': usd_size
        }
        print(f'{index} - Bought Bitcoin for {usd_size}$ at price of {buy_price}$')
        
    if position and row['sell_signal'] is True:
        sell_price = row['close']
        trade_results = (sell_price - position['buy_price']) / position['buy_price']
        balance += trade_results *position['usd_size']
        print(f'{index} - Sold Bitcoin for {balance}$ at price of {sell_price}$({round(trade_results*100,2)}%)')
        position = None
```

```python
print(f"Our final balance is {round(balance,2)}$")
```

```python
# Trading strategy written with Functions

class BacktestCrossMA:
    def __init__(self) -> None:
        self.df = pd.DataFrame()
        
    def load_data(self,path):
        self.df =pd.read_csv(path)
        self.df['date'] = pd.to_datetime(self.df['date'], unit ='ms')
        self.df = self.df.set_index(self.df['date'])
        del self.df['date']
    
    def populate_indicators(self):
        self.df['ma20'] = self.df['close'].rolling(20).mean()
        self.df['ma50'] = self.df['close'].rolling(50).mean()
        
    def populate_signals(self):
        self.df['buy_signal'] = False
        self.df['sell_signal'] = False
        self.df.loc[(self.df['ma20'] > self.df['ma50']), 'buy_signal'] = True
        self.df.loc[(self.df['ma20'] < self.df['ma50']), 'sell_signal'] = True
        
    def run_backtest(self):
        balance = 1000
        position = None
        asset = 'BTC'
        
        for index, row in self.df.iterrows():
            
            if position is None and row['buy_signal']:
                open_price = row['close']
                position = {
                    'open_price': open_price,
                    'usd_size': balance,
                }
                #print(f"{index} - Buy for {balance}$ of {asset} at {open_price}$")
        
            elif position and row['sell_signal']:
                close_price = row['close']
                trade_results = (close_price - position['open_price']) / position['open_price']
                balance = balance + trade_results * position['usd_size']
                position = None
                #print(f"{index} - Sell for {balance}$ of {asset} at {close_price}$")
        
        print(f"Final balance: {balance}$")
        
bt = BacktestCrossMA()
bt.load_data('BNB-USDT.csv')
bt. populate_indicators()
bt.populate_signals()
bt.run_backtest()

bt = BacktestCrossMA()
bt.load_data('BTC-USDT.csv')
bt. populate_indicators()
bt.populate_signals()
bt.run_backtest()
```
